## Dolan junction

```python
matplotlib.use("QtAgg")

# --- configure quickplot behavior ---
set_qp_options(
    new_window=True,     # open a new window
    blocking=True,      # don't halt script execution
    show_ports=True,     # draw ports
    show_subports=True, # don't show ports of references
)
```

```python
dolan_junction_params = DolanJunctionParams()

dolan_junction = DolanJunction()

D = dolan_junction.build_dolan_junction(dolan_junction_params)

qp(D)

D.write_gds(filename='Dolan_Junction_Default.gds')
```

```python
dolan_junction_params = DolanJunctionParams(stress_shape='asym_poly')

dolan_junction = DolanJunction()

D = dolan_junction.build_dolan_junction(dolan_junction_params)

qp(D)

D.write_gds(filename='Dolan_Junction_Asym_Poly.gds', precision=1e-10)
```

Precision can be important when writing to gds! Compare undercuts on these two files in KLayout.
